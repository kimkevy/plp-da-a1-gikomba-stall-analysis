```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100, 2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
  "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
  "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
  "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
  "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
  "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}

# Compute overall statistics
total_sales_sum = sum(daily_sales)
avg_sales = total_sales_sum / len(daily_sales)
max_sales = max(daily_sales)
min_sales = min(daily_sales)

# Find the index of the best and worst days
best_day_idx = daily_sales.index(max_sales)
worst_day_idx = daily_sales.index(min_sales)

# Identify day name and week number (Week 1 for first 7 days, Week 2 for next 7 days)
best_day_name = days[best_day_idx]
best_day_week = 1 if best_day_idx < 7 else 2

worst_day_name = days[worst_day_idx]
worst_day_week = 1 if worst_day_idx < 7 else 2

print(f"Total Sales: KES {total_sales_sum:,.2f}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Highest Daily Sales: KES {max_sales:,.2f} ({best_day_name}, Week {best_day_week})")
print(f"Lowest Daily Sales: KES {min_sales:,.2f} ({worst_day_name}, Week {worst_day_week})")

week_total_sales = sum(daily_sales[:7])
print(f"Week 1 Total Sales: KES {week_total_sales:,.2f}")
print(f"Week 2 Total Sales: KES {sum(daily_sales[7:]):,.2f}")
```

```python
def week_totals(sales):
    """Returns a tuple of (week 1 total, week 2 total) using slicing"""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return (week1, week2)

# Calculate week-on-week growth percentage
w1_total, w2_total = week_totals(daily_sales)
wow_growth = ((w2_total - w1_total) / w1_total) * 100

print(f"Week 1 Total: KES {w1_total:,.2f}")
print(f"Week 2 Total: KES {w2_total:,.2f}")
print(f"Week-on-Week Growth: {wow_growth:+.2f}%")
```

```python
def classify_day(amount, average):
    """Classifies day sales relative to the average"""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"

# Count days per category
day_categories = {}
for sale in daily_sales:
    category = classify_day(sale, avg_sales)
    day_categories[category] = day_categories.get(category, 0) + 1

print("Day Classification Counts:")
for cat, count in day_categories.items():
    print(f"  {cat}: {count} days")
```

```python
# Weekdays (Mon-Fri) indices: 0-4 (Week 1), 7-11 (Week 2)
# Weekends (Sat-Sun) indices: 5-6 (Week 1), 12-13 (Week 2)

weekday_sales = [daily_sales[i] for i in range(14) if i % 7 < 5]
weekend_sales = [daily_sales[i] for i in range(14) if i % 7 >= 5]

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)

weekend_vs_weekday_diff = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend to Weekday Sales Difference: {weekend_vs_weekday_diff:+.2f}%")
```

```python
def check_reorder_and_margins(stock_dict):
    """Returns a list of items to reorder and prints gross margins"""
    to_reorder = []
    margins = {}
    for product, details in stock_dict.items():
        margin = ((details["price"] - details["cost"]) / details["price"]) * 100
        margins[product] = margin
        if details["qty"] <= details["reorder"]:
            to_reorder.append(product)
    return to_reorder, margins

reorder_items, gross_margins = check_reorder_and_margins(stock)

print("Gross Margin Percentages:")
for product, margin in gross_margins.items():
    print(f"  {product}: {margin:.2f}%")

print(f"\nItems needing reorder: {', '.join(reorder_items)}")
```

```python
print("=" * 50)
print("             STALL PERFORMANCE REPORT             ")
print("=" * 50)
print(f"Total 14-Day Sales:       KES {total_sales_sum:,.2f}")
print(f"Average Daily Sales:      KES {avg_sales:,.2f}")
print(f"Week-on-Week Growth:      {wow_growth:+.2f}%")
print("-" * 50)
print(f"Avg Weekday Sales (M-F):  KES {avg_weekday:,.2f}")
print(f"Avg Weekend Sales (S-S):  KES {avg_weekend:,.2f} ({weekend_vs_weekday_diff:+.2f}% higher)")
print("-" * 50)
print("Sales Day Classifications:")
for cat, count in day_categories.items():
    print(f"  - {cat} Days:           {count}")
print("-" * 50)
print("Stock Margin & Reorder Status:")
for product, margin in gross_margins.items():
    reorder_status = "REORDER NOW" if product in reorder_items else "OK"
    qty = stock[product]['qty']
    print(f"  - {product:10} | Margin: {margin:.1f}% | Stock: {qty:2d} kg | Status: {reorder_status}")
print("=" * 50)
```
